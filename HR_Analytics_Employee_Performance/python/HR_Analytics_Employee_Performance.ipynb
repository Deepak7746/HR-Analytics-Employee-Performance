```python
total_employees = df["Employee_ID"].nunique()

high_performers = (df["Performance_Score"] >= 80).sum()

low_performers = (df["Performance_Score"] < 50).sum()

attrition_count = (df["Attrition"] == "Yes").sum()

attrition_rate = attrition_count / total_employees * 100

avg_performance = df["Performance_Score"].mean()

avg_satisfaction = df["Job_Satisfaction"].mean()

avg_training = df["Training_Hours"].mean()

print("Total Employees:", total_employees)
print("High Performers:", high_performers)
print("Low Performers:", low_performers)
print("Attrition Count:", attrition_count)
print("Attrition Rate:", round(attrition_rate, 2), "%")
print("Average Performance:", round(avg_performance, 2))
print("Average Satisfaction:", round(avg_satisfaction, 2))
print("Average Training Hours:", round(avg_training, 2))
```

```python
df_clean = df.copy()

print("Original Shape:", df_clean.shape)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(" ", "_")
)

print(df_clean.columns.tolist())
```

```python
print("Duplicate Employee IDs:",
      df_clean["Employee_ID"].duplicated().sum())
```

```python
df_clean = df_clean.drop_duplicates(subset="Employee_ID")

print("Shape after duplicate removal:", df_clean.shape)
```

```python
missing = df_clean.isnull().sum()

print(missing[missing > 0])
```

```python
text_columns = df_clean.select_dtypes(include="object").columns

for col in text_columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()
```

```python
df_clean[text_columns].head()
```

```python
df_clean["Joining_Date"] = pd.to_datetime(
    df_clean["Joining_Date"],
    errors="coerce"
)

df_clean["Performance_Review_Date"] = pd.to_datetime(
    df_clean["Performance_Review_Date"],
    errors="coerce"
)
```

```python
print(df_clean[["Joining_Date", "Performance_Review_Date"]].dtypes)
```

```python
print(
    "Invalid Age:",
    ((df_clean["Age"] < 18) | (df_clean["Age"] > 70)).sum()
)
```

```python
print(
    "Invalid Performance Score:",
    ((df_clean["Performance_Score"] < 0) |
     (df_clean["Performance_Score"] > 100)).sum()
)
```

```python
print(
    "Invalid Goal Achievement:",
    ((df_clean["Goal_Achievement"] < 0) |
     (df_clean["Goal_Achievement"] > 100)).sum()
)
```

```python
print(
    "Invalid Productivity Score:",
    ((df_clean["Productivity_Score"] < 0) |
     (df_clean["Productivity_Score"] > 100)).sum()
)
```

```python
satisfaction_columns = [
    "Job_Satisfaction",
    "Environment_Satisfaction",
    "Work_Life_Balance",
    "Relationship_Satisfaction"
]

for col in satisfaction_columns:
    invalid = (
        (df_clean[col] < 1) |
        (df_clean[col] > 5)
    ).sum()

    print(col, "Invalid Values:", invalid)
```

```python
categorical_columns = [
    "Gender",
    "Education_Level",
    "Marital_Status",
    "Department",
    "Job_Role",
    "Job_Level",
    "Performance_Rating",
    "Salary_Band",
    "Overtime",
    "Promotion_Status",
    "Attrition",
    "Employee_Status"
]

for col in categorical_columns:
    print("\n", col)
    print(df_clean[col].unique())
```

```python
df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=[
        "Under 25",
        "25-34",
        "35-44",
        "45-54",
        "55+"
    ]
)
```

```python
df_clean["Experience_Group"] = pd.cut(
    df_clean["Years_At_Company"],
    bins=[-1, 2, 5, 10, 20, 100],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)
```

```python
df_clean["Training_Effectiveness"] = (
    df_clean["Training_Hours"] /
    (df_clean["Training_Programs"] + 1)
)
```

```python
df_clean["Promotion_Delay_Flag"] = np.where(
    df_clean["Years_Since_Promotion"] >= 4,
    "Delayed Promotion",
    "Normal"
)
```

```python
df_clean["Absenteeism_Category"] = pd.cut(
    df_clean["Absenteeism_Days"],
    bins=[-1, 3, 7, 15, 100],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Very High"
    ]
)
```

```python
df_clean["Performance_Category"] = pd.cut(
    df_clean["Performance_Score"],
    bins=[-1, 49, 69, 79, 89, 100],
    labels=[
        "Low",
        "Moderate",
        "Good",
        "Very Good",
        "Excellent"
    ]
)
```

```python
df_clean["Employee_Risk"] = np.select(
    [
        (
            (df_clean["Performance_Score"] < 50) &
            (df_clean["Job_Satisfaction"] <= 2)
        ),

        (
            (df_clean["Performance_Score"] < 70) |
            (df_clean["Job_Satisfaction"] <= 3) |
            (df_clean["Overtime"] == "Yes")
        )
    ],
    [
        "High Risk",
        "Medium Risk"
    ],
    default="Low Risk"
)
```

```python
from sklearn.preprocessing import MinMaxScaler

performance_features = [
    "Performance_Score",
    "Goal_Achievement",
    "Productivity_Score",
    "Job_Satisfaction",
    "Training_Hours",
    "Certification_Count"
]

scaler = MinMaxScaler()

scaled_values = scaler.fit_transform(
    df_clean[performance_features]
)

scaled_df = pd.DataFrame(
    scaled_values,
    columns=performance_features,
    index=df_clean.index
)
```

```python
df_clean["Performance_Index"] = (
    scaled_df["Performance_Score"] * 0.30 +
    scaled_df["Goal_Achievement"] * 0.20 +
    scaled_df["Productivity_Score"] * 0.20 +
    scaled_df["Job_Satisfaction"] * 0.10 +
    scaled_df["Training_Hours"] * 0.10 +
    scaled_df["Certification_Count"] * 0.10
) * 100
```

```python
df_clean["Performance_Index"].describe()
```

```python
df_clean["Performance_Segment"] = np.select(
    [
        df_clean["Performance_Index"] >= 80,
        df_clean["Performance_Index"] >= 65,
        df_clean["Performance_Index"] >= 50
    ],
    [
        "High Performer",
        "Solid Performer",
        "Needs Improvement"
    ],
    default="Critical"
)
```

```python
df_clean["Performance_Segment"].value_counts()
```

```python
df_clean["Training_Level"] = pd.cut(
    df_clean["Training_Hours"],
    bins=[-1, 20, 40, 60, 1000],
    labels=[
        "Low Training",
        "Moderate Training",
        "High Training",
        "Very High Training"
    ]
)
```

```python
training_analysis = (
    df_clean.groupby("Training_Level", observed=True)
    .agg(
        Avg_Performance=("Performance_Score", "mean"),
        Avg_Productivity=("Productivity_Score", "mean"),
        Avg_Satisfaction=("Job_Satisfaction", "mean"),
        Employee_Count=("Employee_ID", "nunique")
    )
)

print(training_analysis)
```

```python
promotion_analysis = (
    df_clean.groupby("Promotion_Status")
    .agg(
        Avg_Performance=("Performance_Score", "mean"),
        Avg_Productivity=("Productivity_Score", "mean"),
        Avg_Satisfaction=("Job_Satisfaction", "mean"),
        Avg_Years_Since_Promotion=("Years_Since_Promotion", "mean"),
        Employee_Count=("Employee_ID", "nunique")
    )
)

print(promotion_analysis)
```
